#### 2. Build a silver table from bronze that removes duplicates, fixes data types, and drops clearly invalid rows.

```python
%sql
create schema if not exists cyntexa_dev.silver
```

```python
%sql
select * from cyntexa_dev.bronze.sales_raw
```

```python
from pyspark.sql import functions as F

raw = spark.read.table('cyntexa_dev.bronze.sales_raw')
df = raw.dropDuplicates().drop(F.col("_rescued_data")).dropna()

#always check before saving the data into tables
df.display()
```

```python
df.write.mode('overwrite').saveAsTable('cyntexa_dev.silver.sales_clean')
```

#### 3. Build a gold table that aggregates silver into a business-ready view (e.g., daily revenue by store).

```python
%sql
create schema if not exists cyntexa_dev.gold
```

```python
silver = spark.read.table('cyntexa_dev.silver.sales_clean')
silver.display()
```

```python
from pyspark.sql import functions as F

contribute_df = silver.select(F.sum("total_amount")).collect()[0][0]
silver_df = silver.groupBy("product_id").agg(F.sum("quantity").alias("total_quantity"), F.round(F.sum("total_amount"), 2).alias("total_revenue"), F.round((F.sum("total_amount")/contribute_df)*100, 2).alias("product_contribution")).orderBy("total_revenue", ascending=False)
silver_df.display()
```

```python
silver_df.write.mode('overwrite').saveAsTable('cyntexa_dev.gold.total_revenue_by_products')
```

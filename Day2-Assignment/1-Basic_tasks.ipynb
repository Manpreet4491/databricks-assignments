```python
%sql
create table if not exists cyntexa_dev.sales.orders_raw (order_id INT, order_date DATE, customer_id INT, price DECIMAL(10,2), status STRING)
```

#### 3. Create a view sales.orders_view that selects only completed orders.

```python
%sql 
create view cyntexa_dev.sales.orders_view as 
select * from cyntexa_dev.sales.orders_raw 
where status = 'DELIVERED'
```

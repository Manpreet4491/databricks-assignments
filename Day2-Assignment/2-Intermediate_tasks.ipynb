#### 5. Write a SQL UDF that masks the last 4 digits of a customer_id or email column, and apply it in a SELECT against orders_raw.

```python
%sql
create function if not exists cyntexa_dev.sales.mask_customer_id(customer_id string)
returns string
return concat(substr(customer_id, 1, length(customer_id)-4), repeat('*', 4))
```

```python
%sql
select order_id, order_date, cyntexa_dev.sales.mask_customer_id(customer_id), price, status from cyntexa_dev.sales.orders_raw
```

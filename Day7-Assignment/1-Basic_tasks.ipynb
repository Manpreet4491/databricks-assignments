#### 1. Create a table, make 3 changes to it (insert, update, insert), and use DESCRIBE HISTORY to review the resulting versions.

```python
%sql
create or replace table cyntexa_dev.bronze.simple_table(product_id int, product_name string, product_category string, product_price double)
```

```python
%sql
insert into cyntexa_dev.bronze.simple_table values (1, 'product1', 'category1', 100.00), (2, 'product2', 'category2', 200.00)
```

```python
%sql
update cyntexa_dev.bronze.simple_table set product_price = 150.00 where product_id = 1
```

```python
%sql
insert into cyntexa_dev.bronze.simple_table values (3, 'product3', 'category3', 300.00)
```

```python
%sql
describe history cyntexa_dev.bronze.simple_table
```

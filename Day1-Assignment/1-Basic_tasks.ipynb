#### 1. Difference Between a Data Warehouse, a Data Lake and a Data Lakehouse.

##Data Warehouse:
* Definition - It a the large file system in which we can store files either offline or online.
* Usage - It store the legacy or enterprise level data which is mostly in structured form.
* Architecture - It consists of Databases, Schemas and Tables.

##Data Lake:
* Definition - It refers to a storage which stores data in the form of Documents unlike traditional Databases.
* Usage - It is used to store unorganised data like images, videos, pdfs, models, etc.
* Architecture - It consists of Documents, Collection, Buckets and Key-values.

##Data Lakehouse:
* Definition - It the hybrid form of Data Warehouse and Data Lakehouse which supports both the functionalities.
* Usage - It is used to store both the Structured and Unstructured data like images, tables, documents, models, etc.
* Architecture - It Consists of Unity Catalog, Functions, Models, Tables, Volumes, etc.

#### 3. Write a command to insert or update to see the history of the table.

```sql
%sql
insert into dev.demo.sample_products values (11, 'T-Shirt', 'Clothing', 12.99)
```

```sql
%sql
describe history dev.demo.sample_products
```

```sql
%sql
update dev.demo.sample_products set price = price * 1.1 where price > 50
```

```sql
%sql
describe history dev.demo.sample_products
```

```sql
%sql
update dev.demo.sample_products set name = 'Over Sized T Shirt' where product_id = 11
```

```sql
%sql
describe history dev.demo.sample_products
```
